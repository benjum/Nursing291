## 11. Keyboard shortcut cheat sheet

| Command mode (**Esc**) | | Edit mode (**Enter**) | |
| --- | --- | --- | --- |
| **A** / **B** | add cell above / below | **Shift + Enter** | run, move to next |
| **M** / **Y** | Markdown / Code | **Ctrl + Enter** | run, stay here |
| **D, D** | delete cell | **Alt + Enter** | run, add a cell below |
| **Z** | undo cell action | **Tab** | auto-complete |
| **↑ / ↓** | move between cells | **Shift + Tab** | quick help |
| **0, 0** | restart kernel | **Ctrl + S** | save |

More shortcuts: **Help → Show Keyboard Shortcuts**
