# An Illustrative R Notebook

This notebook shows how to:
1. Make a very simple plot
2. Import data from an Excel file
3. Calculate a statistic
4. Make a plot from the data
5. Save the plot

We will analyze `sample_data.xlsx`, which should be in the same folder as this notebook.

## Setup
We will use the `readxl` package to read Excel files. 

This is already installed in the environment.  If you are running this on another system (such as your local computer), you can run the install line once if you don't have the package.

```r
# install.packages("readxl")   # uncomment and run once if needed
```

```r
library(readxl)                # imports the library
```

## A very simple plot
Plot some made-up numbers.

```r
x <- -10:10
y <- x^3

plot(x, y, 
     main = "A very simple plot", 
     xlab = "x", 
     ylab = "x cubed")
```

## Import data from an Excel file

```r
data <- read_excel("sample_data.xlsx")

head(data)   # look at the first few rows
```

## Calculate a statistic
The average height, and the correlation between height and weight.

```r
mean_height <- mean(data$height_cm)
mean_height
```

```r
correlation <- cor(data$height_cm, data$weight_kg)
correlation
```

## Make a plot from the data

```r
plot(data$height_cm, data$weight_kg,
     main = "Height vs. Weight",
     xlab = "Height (cm)", 
     ylab = "Weight (kg)",
     pch = 19, 
     col = "steelblue")
```

## Save the plot
Open a PNG file, draw the plot into it, then close the file with `dev.off()`.

```r
# 1. Open the PNG device with desired dimensions
png("height_vs_weight3.png", width = 800, height = 600)

# 2. Create your plot
plot(data$height_cm, data$weight_kg,
     main = "Height vs. Weight",
     xlab = "Height (cm)", 
     ylab = "Weight (kg)",
     pch = 19, 
     col = "steelblue")

# 3. Close the device to write the file 
# (not strictly needed in notebooks but good practice)
dev.off()

# 4. Output a message
cat("Plot saved to height_vs_weight.png")
```
