## 1. What is Jupyter?

"Jupyter" refers to three things that work together:

| Piece | What it is | Everyday analogy |
| --- | --- | --- |
| **Notebook** (a `.ipynb` file) | A document that mixes code, the results of that code, and ordinary text | A lab notebook |
| **JupyterLab** | The web application, running in your browser, where you open and edit notebooks and other files | Your desk |
| **Kernel** | The program running behind the scenes that actually executes your code | An R console running in the background |

The name *Jupyter* comes from **Ju**lia, **Pyt**hon, and **R**, the first three languages it supported. The notebook interface is the same for every language; the kernel decides which language runs. **This notebook uses the R kernel**, so every code cell here is R code.

## 5. The two modes: Edit mode and Command mode

JupyterLab has two modes, and **your keyboard does different things in each one**.

|  | **Edit mode** | **Command mode** |
| --- | --- | --- |
| What you're doing | Typing *inside* one cell | Working with the notebook as a whole: adding, deleting, moving cells |
| What you'll see | A blinking cursor, and a blue outline around the cell's text box | No cursor, and a blue outline around the whole cell |
| Typing a letter… | types that letter | **runs a shortcut** (for example, **B** adds a new cell!) |
| How to get there | Press **Enter**, or click inside a cell | Press **Esc**, or click in the margin to the left of a cell |
| Status bar says | *Mode: Edit* | *Mode: Command* |

## 7. The kernel: your notebook's memory

The kernel is a live R session. It remembers every variable you've created, **in the order you ran the cells, not the order they appear on the page.** You can run cells out of order, run the same cell twice, or delete a cell whose variable still lives on in memory. This is the #1 cause of confusing notebook results.

✏️ **Try it:** Run the three cells below in order. Then go back and run the *middle* cell three more times, and run the last cell again. What is `counter` now? Compare the execution counts to see what happened.

````r
counter <- 0
````

````r
counter <- counter + 1
````

````r
counter
````

To list the names of every variable the kernel currently remembers, use `ls()`:

````r
ls()
````

### Restarting the kernel

**Restarting** gives you a fresh R session: every variable and loaded package is forgotten. Your code and text stay in the notebook; only R's memory is wiped. To restart, use **Kernel → Restart Kernel…**, the ⟳ button in the toolbar, or press **0, 0** in Command mode.

The **Kernel** menu has two especially useful variations:

* **Restart Kernel and Clear Outputs of All Cells…** gives you a clean slate.
* **Restart Kernel and Run All Cells…** is the best way to check your work. It proves your notebook runs correctly from top to bottom, in order, with nothing left over from earlier experiments. **Do this before you submit or share a notebook.** (If a cell has an error, Run All stops there, which also makes it a quick way to find problems.)

✏️ **Try it:** Restart the kernel now, then run the cell below. You'll get the error `object 'counter' not found`, because R has forgotten it. That's expected!

````r
counter
````

If a cell runs for a long time or seems stuck, **interrupt** it with the ■ (Stop) button in the toolbar, **Kernel → Interrupt Kernel**, or **I, I** in Command mode. While code is running, the kernel indicator shows a filled circle ● and the cell shows `[*]`.

### Plots appear inside the notebook

There's no separate plot window. The plot appears under the cell. You can set the plot size (in inches) with `options(repr.plot.width = ..., repr.plot.height = ...)`; the setting lasts until you change it or restart the kernel.

### Warnings and errors

When R wants your attention, the output appears on a **light red background**. Don't panic:

* A **warning** means R finished, but something looks suspicious.
* An **error** means R stopped. The **last line** of the message is usually the most helpful part.

An error only stops that one cell. Fix the code and run the cell again.

````r
log(-1)   # a warning: the logarithm of a negative number is NaN ("Not a Number")
````

## 13. Saving, exporting, and finishing up

### Saving

* JupyterLab autosaves every couple of minutes, but don't rely on it. Save often with **Ctrl + S** or **File → Save Notebook**. A dot ● on the notebook's tab means there are unsaved changes.
* Each save also creates a *checkpoint*. If you make a mess, **File → Revert Notebook to Checkpoint…** takes you back to your last save.
* To rename a notebook, right-click its tab (or the file in the file browser) and choose **Rename…**.

### Exporting and sharing

* **File → Download** saves the `.ipynb` file to your computer.
* **File → Save and Export Notebook As** converts the notebook to other formats, such as **HTML**, which anyone can open in a web browser without Jupyter. (PDF export may need extra software on the server; if it fails, export to HTML and print that to PDF from your browser.)

### Shutting down

* **Closing a notebook's tab does *not* stop its kernel.** The R session keeps running and using memory. Use **File → Close and Shut Down Notebook** (Ctrl + Shift + Q) instead.
* The **Running Terminals and Kernels** tab in the left sidebar lists everything that's still running, with a **Shut Down All** button.
* **On a shared server (JupyterHub):** when you're finished for the day, save, then use **File → Shut Down**. This frees up resources for everyone else.
